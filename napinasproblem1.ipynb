```python
FILENAME = "records.txt"

def compute_final_grade(quiz, exam):
    return round((quiz * 0.4) + (exam * 0.6), 2)

def load_records():
    records = []
    try:
        with open(FILENAME, "r") as file:
            for line in file:
                data = line.strip().split(",")
                if len(data) == 6:
                    student = {
                        "id": data[0],
                        "name": data[1],
                        "course": data[2],
                        "quiz": float(data[3]),
                        "exam": float(data[4]),
                        "final": float(data[5])
                    }
                    records.append(student)
    except FileNotFoundError:

        pass
    return records


def save_records(records):
    with open(FILENAME, "w") as file:
        for s in records:
            file.write(f"{s['id']},{s['name']},{s['course']},{s['quiz']},{s['exam']},{s['final']}\n")



def add_student(records):
    student_id = input("Enter Student ID: ")

    for s in records:
        if s["id"] == student_id:
            print("Student ID already exists!")
            return

    name = input("Enter Fullname: ")
    course = input("Enter Course: ")

    try:
        quiz = float(input("Enter Quiz Score: "))
        exam = float(input("Enter Exam Score: "))
    except ValueError:
        print("Scores must be numeric.")
        return

    final = compute_final_grade(quiz, exam)

    student = {
        "id": student_id,
        "name": name,
        "course": course,
        "quiz": quiz,
        "exam": exam,
        "final": final
    }

    records.append(student)
    save_records(records)
    print("Student record added successfully!")


def view_records(records):
    if not records:
        print("No records found.")
        return

    print("\n--- Student Records ---")
    for s in records:
        print(f" Student ID: {s['id']} \n Name: {s['name']} \n Course: {s['course']} \n "
              f"Quiz: {s['quiz']} \n Exam: {s['exam']} \n Final: {s['final']}\n")
    print()


def search_student(records):
    student_id = input("Enter Student ID to search: ")
    for s in records:
        if s["id"] == student_id:
            print("\nStudent Found:")
            print(s)
            return
    print("Student not found.")


def update_scores(records):
    student_id = input("Enter Student ID to update: ")

    for s in records:
        if s["id"] == student_id:
            try:
                s["quiz"] = float(input("Enter new Quiz Score: "))
                s["exam"] = float(input("Enter new Exam Score: "))
            except ValueError:
                print("Scores must be numeric.")
                return

            s["final"] = compute_final_grade(s["quiz"], s["exam"])
            save_records(records)
            print("Scores updated successfully!")
            return

    print("Student not found.")


def compute_average(records):
    if not records:
        print("No records available.")
        return

    total = sum(s["final"] for s in records)
    average = round(total / len(records), 2)
    print(f"Class Average Final Grade: {average}")



def main():
    records = load_records()

    while True:
        print("\n- Student Enrollment and Grading System -")
        print("(1) Add Student Record")
        print("(2) View All Records")
        print("(3) Search Student by ID")
        print("(4) Update Student Scores")
        print("(5) Compute Class Average")
        print("(6) Exit")

        choice = input("Enter your option: ")

        if choice == "1":
            add_student(records)
        elif choice == "2":
            view_records(records)
        elif choice == "3":
            search_student(records)
        elif choice == "4":
            update_scores(records)
        elif choice == "5":
            compute_average(records)
        elif choice == "6":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid menu selection. Please try again.")


main()
```
